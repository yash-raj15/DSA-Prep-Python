````python
n=[2,4,1,3,5]
j=len(n)
for i in range(j//2):
    temp=n[i]
    n[i]=n[j-1-i]
    n[j-1-i]=temp
print(n)
````

````python
n=[2,4,1,3,5]
l=0
r=len(n)-1
while l<r:
    n[l],n[r]=n[r],n[l]
    l+=1
    r-=1
print(n)
````

````python
# 41. Sort ascending

n=[2,4,1,3,5]
l=len(n)


for i in range(l-1):
    for j in range(l-1-i):
        if n[j]>n[j+1]:
            n[j],n[j+1]=n[j+1],n[j]

print(n)



````

````python
# 42. Sort descending

n=[2,4,1,3,5]
l=len(n)

for i in range(l-1):
    for j in range(l-1-i):
        if n[j]>n[j+1]:
            n[j],n[j+1]=n[j+1],n[j]
print(n[::-1])
````

````python
# 43. Find second largest
# 

n=[2,4,1,3,5]
l=len(n)

for i in range(l-1):
    for j in range(l-1-i):
        if n[j]>n[j+1]:
            n[j],n[j+1]=n[j+1],n[j]
print(n[-2])
````
