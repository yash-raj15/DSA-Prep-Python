# Function With Return

Given an integer n, You need to use the return keyword to return twice the value of n.

Examples:

Input: n = 2
Output: 4
Explanation: 2 * 2 = 4
Input: n = 4
Output: 8
Explanation: 2 * 4 = 8

class Solution:
    def returnValueFunction(self, n):
        return 2*n
